```python
# My partner did question 2. The task was to find all the pathways from the root to every single leaf of a binary tree. Basically, 
# each pathway provides the path to one of the leafs and combined is the total number of pathways of that binary tree. 
```


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


```python
# My new  Example:
# Input:
root = [2, 3, 5, 1, 7, 9, 4]
# Output:
[[2,3,1],[2,3,7],[2,5,9],[2,5,4]]

## Explanation of Partner's example 2 
# Example 2:
# Input:
root = root = [3, 9, 20, 15, 7]
# Output:
[[3, 9], [3, 20, 15], [3, 20, 7]]

# 1. the first number is the root (3)
# 2. the next number is one of the branches from that root (9)
# 3. This is the other branch from the root (20)
# 4. 15 and 7 seem to be the binary branches off the number 20 
```


-   Copy the solution your partner wrote. 


```python

def bt_path(root: TreeNode) -> List[List[int]]:
    # List to store all the paths
    paths = []
    
    # function to perform DFS traversal
    def dfs(node, current_path):
        if node is None:
            return
        
        # Add the current node's value to the path
        current_path.append(node.val)
        
        # If it's a leaf node, store the current path
        if node.left is None and node.right is None:
            paths.append(list(current_path))  
        
        # Recursively visit the left and right child nodes
        dfs(node.left, current_path)
        dfs(node.right, current_path)
        
        # Backtrack: remove the current node from the path before going back
        current_path.pop()
    
    # Start DFS traversal from the root node
    dfs(root, [])
    
    return paths

# Example 1:
# Input:
root = TreeNode(5)
root.left = TreeNode(4)
root.right = TreeNode(8)
root.left.left = TreeNode(11)
root.left.left.left = TreeNode(7)
root.left.left.right = TreeNode(2)
root.right.left = TreeNode(13)
root.right.right = TreeNode(4)
root.right.right.left = TreeNode(5)
root.right.right.right = TreeNode(1)


print(f"Input: {root}")

print(f"Output: {bt_path(root)}")
```


-   Explain why their solution works in your own words.


```python
# This works because the solution uses the following: 
# 1. the depth first search (dfs) function which starts at a root and goes further down the nodes until it reaches an end. Then
# retraces steps back to the top and keeps doing so until all paths have been mapped. 
# 2. the solution uses the function and at each step appends the value found to a list called path. 
# 3. once it reaches the end, then the list is stored and the function starts at the root again
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Time complexity
# Since this is a depth first search and we visit each node, then the time complexity is dependant on the number of nodes, n - O(n)

# space complexity 
# since this is recursive and each step of the way is stored, the depth of the tree affects the space. If one of the pathways is longer, though
# that will also affect the space complexity. Worst case scenario is if the tree is just one long path and that space complexity will relect 
# the number of nodes in that case O(n).
```
